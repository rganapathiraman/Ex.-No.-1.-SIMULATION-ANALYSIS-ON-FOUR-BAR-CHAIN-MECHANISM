#<h3> Task 1</h3>

Create a quantum circuit with two qubits, say $ q[1] $ and $ q[0] $ in the reading order of Qiskit.

We start in quantum state $ \ket{01} $:
- set the state of $ q[1] $ to $ \ket{0} $, and
- set the state of $ q[0] $ to $ \ket{1} $.

Apply Hadamard to both qubits.

Apply CNOT operator, where the controller qubit is $ q[1] $ and the target qubit is $ q[0] $.

Apply Hadamard to both qubits.

Measure the outcomes.

```python
# Install Qiskit and Aer simulator (new modular versions)
!pip install qiskit qiskit-aer -q

# Import the required modules
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# --- Create quantum circuit with 2 qubits and 2 classical bits ---
qc = QuantumCircuit(2, 2)

# Step 1: Prepare |01⟩
qc.x(0)  # Set q[0] = |1>, q[1] remains |0>

# Step 2: Apply Hadamard to both qubits
qc.h(0)
qc.h(1)

# Step 3: Apply CNOT (control = q[1], target = q[0])
qc.cx(1, 0)

# Step 4: Apply Hadamard again to both qubits
qc.h(0)
qc.h(1)

# Step 5: Measure both qubits
qc.measure([1, 0], [1, 0])

# Draw the circuit
print("Quantum Circuit:")
display(qc.draw('mpl'))

# Use AerSimulator backend
sim = AerSimulator()

# Run the simulation
job = sim.run(qc, shots=1024)
result = job.result()
counts = result.get_counts()

# Display results
print("\nMeasurement outcomes:")
print(counts)
plot_histogram(counts)
```

<h3> Task 2 </h3>

Create a circuit  with 7 qubits, say $ q[6],\ldots,q[0] $ in the reading order of Qiskit.

Set the states of the top six qubits to $ \ket{0} $.

Set the state of the bottom qubit to $ \ket{1} $.

Apply Hadamard operators to all qubits.

Apply CNOT operator ($q[1]$,$q[0]$)
<br>
Apply CNOT operator ($q[4]$,$q[0]$)
<br>
Apply CNOT operator ($q[5]$,$q[0]$)

Apply Hadamard operators to all qubits.

Measure all qubits.

For each CNOT operator, is there a phase-kickback effect?

```python
# import all necessary objects and methods for quantum circuits
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit #for qiskit version 2.x
from qiskit_aer import Aer #for qiskit version 2.x
# Install latest Qiskit + Aer simulator
!pip install qiskit qiskit-aer -q

# Import modules
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# --- Create a 7-qubit quantum circuit with 7 classical bits ---
qc = QuantumCircuit(7, 7)

# Step 1: Prepare |0000001⟩  (q[6]..q[1]=|0>, q[0]=|1>)
qc.x(0)  # Flip bottom qubit to |1>

# Step 2: Apply Hadamard to all qubits
qc.h(range(7))

# Step 3: Apply CNOT operations
qc.cx(1, 0)   # Control q[1], target q[0]
qc.cx(4, 0)   # Control q[4], target q[0]
qc.cx(5, 0)   # Control q[5], target q[0]

# Step 4: Apply Hadamard again to all qubits
qc.h(range(7))

# Step 5: Measure all qubits
qc.measure(range(7), range(7))

# Display the circuit
print("Quantum Circuit:")
display(qc.draw('mpl'))

# Simulate with AerSimulator
sim = AerSimulator()
job = sim.run(qc, shots=1024)
result = job.result()
counts = result.get_counts()

# Display measurement outcomes
print("\nMeasurement outcomes:")
print(counts)
plot_histogram(counts)

```
